## Conclusion

Attrition is 16.1% (237 left, 1,233 stayed). The classes are imbalanced, so accuracy would mislead in a model.

The workforce is junior and male-skewed. About 73% are at job level 1 or 2, and 60% are male.

R&D is 65% of staff, Sales 30% and HR 4%. HR is too small for reliable percentages.

28% work overtime, and 19% travel frequently.
Satisfaction is low for a large minority. About 39% score 1 or 2 on job satisfaction, and 29% on work-life balance.          
Performance ratings are almost useless. 85% are rated 3, none are below 3.
Promotion is uneven. Most were promoted recently, but a small group has waited 10+ years.
